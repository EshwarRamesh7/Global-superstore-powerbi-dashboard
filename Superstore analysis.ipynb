```python
# =============================================================================
# FUNCTION TO CREATE QUARTILE-BASED CATEGORIES
# =============================================================================

def percentile_category(series, labels):
    """
    Categorizes a numeric column into 4 equal-sized groups (quartiles).

    Parameters
    ----------
    series : pandas Series
        Numeric column.

    labels : list
        Four labels for the quartiles.

    Returns
    -------
    pandas Series
        Categorized column.
    """

    return pd.qcut(
        series.rank(method="first"),
        q=4,
        labels=labels
    )
```
